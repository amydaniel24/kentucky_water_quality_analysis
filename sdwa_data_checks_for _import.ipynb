```python
from pathlib import Path
import pandas as pd

project_root = Path("..") if Path.cwd().name == "notebooks" else Path(".")
source_file = project_root / "datasets" / "SDWA_ky" / "ky_violations_enforcement.csv"
output_file = project_root / "datasets" / "SDWA_ky" / "ky_violations_import_small.csv"

use_cols = [
    "VIOLATION_ID",
    "PWSID",
    "CONTAMINANT_CODE",
    "VIOLATION_CODE",
    "VIOLATION_CATEGORY_CODE",
    "IS_HEALTH_BASED_IND",
    "NON_COMPL_PER_BEGIN_DATE",
    "NON_COMPL_PER_END_DATE"
]

viol_small = pd.read_csv(source_file, usecols=use_cols, low_memory=False)


viol_small = viol_small[viol_small["PWSID"].astype(str).str.startswith("KY")].copy()
viol_small = viol_small.dropna(subset=["VIOLATION_ID", "PWSID", "CONTAMINANT_CODE"])

viol_small["CONTAMINANT_CODE"] = (
    pd.to_numeric(viol_small["CONTAMINANT_CODE"], errors="coerce")
    .dropna()
)

viol_small = viol_small.dropna(subset=["CONTAMINANT_CODE"]).copy()
viol_small["CONTAMINANT_CODE"] = viol_small["CONTAMINANT_CODE"].astype(int).astype(str).str.zfill(4)


viol_small["VIOLATION_ID"] = viol_small["VIOLATION_ID"].astype(str).str.strip()
viol_small["PWSID"] = viol_small["PWSID"].astype(str).str.strip()
viol_small["VIOLATION_CODE"] = viol_small["VIOLATION_CODE"].astype(str).str.strip()
viol_small["VIOLATION_CATEGORY_CODE"] = viol_small["VIOLATION_CATEGORY_CODE"].astype(str).str.strip()
viol_small["IS_HEALTH_BASED_IND"] = viol_small["IS_HEALTH_BASED_IND"].astype(str).str.strip()

viol_small.to_csv(output_file, index=False)

print("Saved:", output_file)
print("Shape:", viol_small.shape)
print(viol_small.head())
```
